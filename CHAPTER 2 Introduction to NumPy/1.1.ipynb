# The Basics of NumPy Array

```python
np.random.seed(0)
```

```python
x1 = np.random.randint(10, size=6) # One-Dimensional Array
print(x1)
print(type(x1))
print(x1.size)
print(x1.shape)
print(x1.ndim)
```

```python
x2 = np.random.randint(10, size=(3,4))  # Two-Dimensions array
print(x2)
print(type(x2))
print(x2.size)
print(x2.shape)
print(x2.ndim)
```

```python
x3 = np.random.randint(10, size=(3,4,5))  # Two-Dimensions array
print(x3)
print(type(x3))
print(x3.size)
print(x3.shape)
print(x3.ndim)
```

```python
print("itemsize: ", x3.itemsize, "bytes")
print("nbytes: ", x3.nbytes, "bytes")
```

## Reshaping of Arrays

```python
grid = np.arange(1, 10).reshape((3,3))
print(grid)
```

```python
x = np.array([1,2,3])
x
```

```python
# row vector vie reshape
x.reshape((1,3))
```

```python
# row vector via newaxis
x[np.newaxis,:]
```

```python
# column vector via reshape
x.reshape((3, 1))
```

```python
 # column vector via newaxis
x[:, np.newaxis]
```
