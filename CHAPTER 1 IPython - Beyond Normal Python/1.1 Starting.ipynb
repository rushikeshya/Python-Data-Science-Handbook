## Controlling Exceptions: %xmode

```python
def func1(a,b):
    return a/b

def func2(x):
    a = x
    b = x-1
    return func1(a,b)

func2(1)
```

#### %xmode
-  Using the %xmode magic function (short for exception mode),
- we can change what information is printed.
%xmode takes a single argument, the mode, and there are three possibilities: 
    - Plain,
    - Context, 
    - and Verbose. 
    
The default is Context, and gives output like that just shown.
Plain is more compact and gives less information:



```python
%xmode Plain
```

```python
func2(1)
```

```python
## Verbose %xmode 
## The Verbose mode adds some extra information, including the arguments to any functions that are called:
%xmode Verbose
```

```python
func2(1)
```

```python
def sum_of_lists(N):
    total = 0
    for i in range(5):
        L = [j ^ (j >> i) for j in range(N)]
        total += sum(L)
    return total
```

```python
%prun sum_of_lists(1000000)
```
